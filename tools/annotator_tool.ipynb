```python
"""
Ground Truth annotator to generate an annotator.html
simply click on the ground truth images, then click Export to copy a list of IDs
"""

import json

with open("paintings_metadata.json", "r") as f:
    meta = json.load(f)

print(f"Generating annotator page {len(meta)} paintings in total...")

cards = []
for pid, m in meta.items():
    t = m.get("title", "").replace('"', '&quot;').replace("<", "&lt;")[:55]
    mk = m.get("maker", "").replace('"', '&quot;').replace("<", "&lt;")
    img = m.get("img_url", "")
    cards.append(
        f'<div class="c" data-id="{pid}" onclick="tog(this)">'
        f'<img src="{img}" loading="lazy" onerror="this.style.display=\'none\'"/>'
        f'<div class="ck">✓</div>'
        f'<div class="nfo"><b>{t}</b><br><span>{mk}</span></div></div>'
    )

html_content = """<!DOCTYPE html>
<html><head><meta charset="UTF-8">
<meta name="viewport" content="width=device-width,initial-scale=1">
<title>Ground Truth Annotation</title>
<style>
*{margin:0;padding:0;box-sizing:border-box}
body{font-family:system-ui;background:#f5f3f0}
.bar{position:sticky;top:0;z-index:10;background:#fff;padding:12px 20px;
  border-bottom:1px solid #ddd;display:flex;align-items:center;gap:14px;flex-wrap:wrap}
.bar h1{font-size:18px}
.bar .st{font-family:monospace;font-size:14px;color:#555}
.bar .st b{color:#2e7d52;font-size:16px}
.bar button{padding:8px 18px;border-radius:6px;border:none;font-size:13px;
  font-weight:600;cursor:pointer}
.bex{background:#1a3a4a;color:#fff}
.bex:hover{background:#254f63}
.bcl{background:#e5e0d8;color:#333}
.hint{padding:8px 20px;font-size:12px;color:#888;background:#fff;border-bottom:1px solid #eee}
.g{display:grid;grid-template-columns:repeat(auto-fill,minmax(170px,1fr));gap:8px;padding:16px 20px}
.c{position:relative;cursor:pointer;border-radius:6px;overflow:hidden;
  border:3px solid transparent;transition:border-color .15s;background:#e8e4de}
.c:hover{box-shadow:0 2px 8px rgba(0,0,0,.12)}
.c.sel{border-color:#2e7d52;background:#f0faf4}
.c.sel .ck{display:flex}
.c img{width:100%;aspect-ratio:1;object-fit:cover;display:block}
.ck{display:none;position:absolute;top:6px;right:6px;width:28px;height:28px;
  border-radius:50%;background:#2e7d52;color:#fff;font-size:15px;font-weight:700;
  align-items:center;justify-content:center;box-shadow:0 1px 4px rgba(0,0,0,.3)}
.nfo{padding:5px 7px;font-size:10px;line-height:1.3;color:#333}
.nfo span{color:#999;font-size:9px}
#out{display:none;margin:16px 20px;padding:16px;background:#fff;border:1px solid #ddd;
  border-radius:8px;font-family:monospace;font-size:11px;white-space:pre-wrap;
  max-height:400px;overflow-y:auto;line-height:1.8;user-select:all}
#cm{display:none;color:#2e7d52;font-weight:600;font-size:13px}
</style></head><body>

<div class="bar">
  <h1>🏺 Ground Truth</h1>
  <div class="st">selected: <b id="cnt">0</b> / """ + str(len(meta)) + """</div>
  <button class="bex" onclick="exp()">Export IDs</button>
  <button class="bcl" onclick="clr()">Clear All</button>
  <span id="cm">✓ Copied!</span>
</div>
<div class="hint">
  Click paintings that contain blue-and-white ceramics (porcelain, Delft, kraak).
  Selected paintings get a green border. Click again to deselect. When done, click Export.
</div>

<div class="g">
""" + "\n".join(cards) + """
</div>

<div id="out"></div>

<script>
function tog(el){
  el.classList.toggle('sel');
  document.getElementById('cnt').textContent=
    document.querySelectorAll('.c.sel').length;
}
function clr(){
  document.querySelectorAll('.c.sel').forEach(e=>e.classList.remove('sel'));
  document.getElementById('cnt').textContent='0';
  document.getElementById('out').style.display='none';
}
function exp(){
  var ids=[];
  document.querySelectorAll('.c.sel').forEach(e=>ids.push(e.dataset.id));
  if(!ids.length){alert('Nothing selected!');return;}

  // Show ID list
  var box=document.getElementById('out');
  var py = 'GROUND_TRUTH = {\\n' + ids.map(i=>'    "'+i+'",').join('\\n') + '\\n}';
  box.textContent = py;
  box.style.display='block';

  // Copy to clipboard
  navigator.clipboard.writeText(py).then(function(){
    var cm=document.getElementById('cm');
    cm.style.display='inline';
    setTimeout(function(){cm.style.display='none';},2000);
  });
}
</script>
</body></html>"""

with open("annotator.html", "w", encoding="utf-8") as f:
    f.write(html_content)

print("✅ annotator.html is created")
print("📥 Download it to open in a browser")
print("   click on ground-truth images → click Export → auto-copy GROUND_TRUTH ID list")
```
